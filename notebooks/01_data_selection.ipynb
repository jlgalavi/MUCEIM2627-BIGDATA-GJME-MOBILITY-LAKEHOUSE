Select a level, date range, and folder. This exploration preserves fields as text and displays only a sample; Silver transformations will be designed in Sprint 1.
The initial selection uses GAU data from 9 to 15 March 2022 because the supplied material contains that complete week. The final project reference year remains 2023.

```python
from pathlib import Path
import sys

repo = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "scripts/datasets.py").is_file()), None)
if repo is None:
    raise RuntimeError("Open the notebook from the mobility-lakehouse or notebooks directory.")
if str(repo) not in sys.path:
    sys.path.insert(0, str(repo))
from scripts.datasets import get_files, load_config, inventory
print("Python:", sys.executable)
inventory(load_config(), None)
```

```python
# None uses the value from config/datasets.toml or datasets.local.toml.
LEVEL = None  # "gau", "distritos", "municipios", or "calidad"
START = None  # "2022-03-09"
END = None    # "2022-03-15"
DATA_ROOT = None  # path to the folder containing GAU, DISTITOS, etc.
files = get_files(LEVEL, START, END, DATA_ROOT)
print(f"Selected {len(files)} files")
for path in files:
    print(path)
```

```python
import duckdb

print("DuckDB:", duckdb.__version__)
with duckdb.connect() as con:
    sample = con.execute(
        "SELECT * FROM read_csv(?, delim='|', all_varchar=true) LIMIT 10",
        [[str(path) for path in files]],
    ).fetchdf()
display(sample)
```
